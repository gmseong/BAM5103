```python
## As always, import pandas as ...

import pandas as pd
```

```python
## Read CSV from GitHub url: https://github.com/AlexTheAnalyst/PandasYouTubeSeries/blob/main/Flavors.csv?raw=true

df = pd.read_csv("https://github.com/AlexTheAnalyst/PandasYouTubeSeries/blob/main/Flavors.csv?raw=true")
```

```python
## Explore the data
df.head()
df.shape
df.info
df
## Success? Commit changes to the repository!
```

```python
## Let's group by Base Flavor

df.groupby("Base Flavor").mean(numeric_only=True)
```

```python
## Make this a new datafram
grouped_df = df.groupby("Base Flavor")

## Success? Commit changes to the repository!
```

```python
## Let's count the number of flavors for each Base Flavor
grouped_df.count()
```

```python
## Let's get the means for each Base Flavor
grouped_df = df.groupby("Base Flavor")
grouped_df.mean(numeric_only=True)

## Success? Commit changes to the repository!
```

```python
## Let's try that in one line
df.groupby("Base Flavor").mean(numeric_only=True)
```

```python
## Let's discover some popular aggregate statistics like count, sum, min, max. Notice that text columns will appear with min and max as well

grouped_df.min()


## Success? Commit changes to the repository!
```

```python
grouped_df.sum(numeric_only=True)
```

```python
grouped_df.agg(['count', 'sum', 'min', 'max'])
```

```python
df.columns
```

```python
## Let's try more advanced aggregation with the `agg` function
## We have to specify which columns we want to aggregate and which functions to apply to them via a dictionary

grouped_df.agg({
    "Flavor Rating": ["mean", "min", "max"],
    "Texture Rating": ["mean", "min", "max"],
    "Total Rating": ["sum", "count"]
})

df

## Success? Commit changes to the repository!
```

```python
## Let's group on multiple columns
## Check the dataframe again and group by Base Flavor and Liked (and show mean as value)
## Per Base Flavor, what is the average Flavor Rating among Liked 'yes' 'no'

df.groupby(["Base Flavor", "Liked"]).mean(numeric_only=True)

## Grouped the original dataframe (df) by Base Flavor and Liked. Then, calculated the mean. Got an error, becasue of strings. The, added the

multigrouped_df = df.groupby(['Base Flavor', 'Liked'])
```

```python
# Do the multiple aggregation for Base Flavor and Liked ('mean', 'max', 'count', 'sum')

df.groupby(["Base Flavor", "Liked"]).agg({
    "Flavor Rating": ["mean", "min", "max"],
    "Texture Rating": ["mean", "min", "max"],
    "Total Rating": ["sum", "count"]
})
```

```python
df.groupby(['Base Flavor', 'Liked']).agg({
    'Flavor Rating': ['mean', 'max', 'count', 'sum']
})
```
