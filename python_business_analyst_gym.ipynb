# Mini business case — Coffee Shop Analytics

You are advising a small coffee shop. Its point-of-sale system exported six orders. Management wants five answers:

1. What is total revenue?
2. How many orders were placed?
3. What is average order value?
4. What was the most frequently sold product?
5. What was revenue by product?

### Suggested workflow

- Write a plain-language plan before coding.
- Solve one question at a time and print intermediate results.
- Check whether your answers reconcile: product revenues must add to total revenue.
- Prepare a 60-second management summary.

```python
orders = [
    {"item": "Latte", "price": 5},
    {"item": "Espresso", "price": 3},
    {"item": "Espresso", "price": 3},
    {"item": "Latte", "price": 5},
    {"item": "Tea", "price": 4},
    {"item": "Tea", "price": 4},
    {"item": "Latte", "price": 5},
    {"item": "Tea", "price": 4},
    {"item": "Latte", "price": 5},
    {"item": "Tea", "price": 4},
]
```

### Step 1 — Write your plan

In your own words, describe how you will calculate each answer. Which collection would help you accumulate revenue by product?

```python
# TODO: calculate the five requested metrics.
case_total_revenue = None
case_order_count = None
case_average_order_value = None
case_most_frequent_product = None
case_revenue_by_product = {}
```

<details><summary><strong>Hints</strong></summary><ul><li>A total can be built with <code>sum(...)</code>.</li><li><code>Counter</code> can count item names.</li><li>For revenue by product, start with an empty dictionary. For each order, add its price to the current total; <code>dictionary.get(key, 0)</code> is useful.</li></ul></details>

```python
# Reference solution
case_total_revenue = sum(order["price"] for order in orders)
case_order_count = len(orders)
case_average_order_value = case_total_revenue / case_order_count

item_counts = Counter(order["item"] for order in orders)
case_most_frequent_product = item_counts.most_common(1)[0][0]

case_revenue_by_product = {}
for order in orders:
    item = order["item"]
    case_revenue_by_product[item] = case_revenue_by_product.get(item, 0) + order["price"]

print("COFFEE SHOP REPORT")
print("------------------")
print(f"Total revenue: ${case_total_revenue:.2f}")
print(f"Orders: {case_order_count}")
print(f"Average order value: ${case_average_order_value:.2f}")
print(f"Most frequent product: {case_most_frequent_product}")
print("Revenue by product:")
for item, item_revenue in case_revenue_by_product.items():
    print(f"  {item}: ${item_revenue:.2f}")

assert sum(case_revenue_by_product.values()) == case_total_revenue
print("Reconciliation check passed.")
```

# 6. Wrap-up challenge (15 minutes)

Without copying an earlier report, turn the quarterly values below into a compact KPI report. Your report must include total sales, average sales, the strongest quarter, and whether annual sales met the target.

```python
quarterly_sales = {"Q1": 120_000, "Q2": 135_000, "Q3": 128_000, "Q4": 155_000}
annual_target = 525_000

# TODO: build the final KPI report.
quarterly_sales = {
    "Q1": 120_000,
    "Q2": 135_000,
    "Q3": 128_000,
    "Q4": 155_000
}

annual_target = 525_000

total_sales = sum(quarterly_sales.values())

average_sales = total_sales / len(quarterly_sales)

strongest_quarter = max(quarterly_sales, key=quarterly_sales.get)

target_met = total_sales >= annual_target
```

```python
## Then complete the report with the following metrics:

print("ANNUAL KPI REPORT")
print("-----------------")
print(f"Total sales: {total_sales:,}")
print(f"Quarterly average: {average_sales:,.0f}")
print(f"Strongest quarter: {strongest_quarter}")
print(f"Target met: {'Yes' if target_met else 'No'}")
```
